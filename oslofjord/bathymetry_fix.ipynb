```python
import numpy as np
import xarray as xr
```

```python
ds_in = xr.open_dataset("/home/sia/FjordsSim_data/oslofjord/OF_inner_105to232_bathymetry_v2.nc")
```

```python
ds_in
```

```python
ds_in["h"] = ds_in.h.where(ds_in.h < 0, np.nan)
```

```python
ds_in.h.plot(figsize=(10, 15))
```

```python
def fill_surrounded_nans(arr):
    # Make a copy to avoid modifying original array
    result = arr.copy()
    rows, cols = arr.shape

    for i in range(1, rows - 1):
        for j in range(1, cols - 1):
            if np.isnan(arr[i, j]):
                west  = arr[i, j - 1]
                north = arr[i - 1, j]
                east  = arr[i, j + 1]
                south = arr[i + 1, j]

                # Check if all 4 neighbors are NOT NaN
                if not np.isnan(west) and not np.isnan(north) and not np.isnan(east) and not np.isnan(south):
                    result[i, j] = np.mean([west, north, east, south])
    
    return result
```

```python
ds_in.h.values = np.nan_to_num(fill_surrounded_nans(ds_in.h.values), nan=10)
```

```python
ds_in.h.plot(figsize=(10, 15))
```
