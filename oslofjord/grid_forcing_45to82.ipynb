```python
import os
from pathlib import Path

import xarray as xr
```

This constructs a forcing file for the 45 to 82 grid from the 105 to 232 forcing file.
So, first, set up the original 105 to 232 forcing file accordingly.

```python
ds_forcing = xr.open_dataset(
    os.path.join(Path.home(), "FjordsSim_data", "oslofjord", "OF_inner_105to232_forcing_v2.nc")
)
ds_in = xr.open_dataset("/home/sia/FjordsSim_data/oslofjord/OF_inner_105to232_bathymetry_v2.nc")
```

```python
ds_out = ds_in.sel(lat=slice(59.6, None), lon=slice(10.45, 10.8))
ds_out["z_faces"] = ds_out.z_faces[6:]
```

```python
ds_forcing_out = ds_forcing.sel(
    Ny=slice(59.6, None), Ny_faces=slice(59.6, None), Nx=slice(10.45, 10.8), Nx_faces=slice(10.445, 10.81)
)
ds_forcing_out = ds_forcing_out.isel(Nz=slice(6, None))
```

```python
ds_day = ds_forcing_out.sel(time="2024-02-29")
ds_day
```

```python
ds_day.u.isel(time=-1, Nz=-1).where(lambda x: x != 0).plot(vmin=-0.5, vmax=0.5)
```

```python
ds_day.u_lambda.isel(time=-1, Nz=-1).where(lambda x: x != 0).plot()
```
